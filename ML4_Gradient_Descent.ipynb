# Gradient Descent
Find the local minimum of y = (x + 3)² starting from x = 2.

### Step 1: Function and its derivative

```python
import matplotlib.pyplot as plt

def f(x):  return (x + 3) ** 2      # function
def df(x): return 2 * (x + 3)       # derivative (slope)
```

### Step 2: Gradient descent loop
Rule: `x_new = x_old - learning_rate * slope`

```python
x = 2             # starting point
lr = 0.1          # learning rate (step size)
precision = 1e-5  # stop when step becomes this small
path = [x]        # store every x for plotting

for i in range(100):                 # max 100 iterations
    x_new = x - lr * df(x)           # move opposite to slope
    path.append(x_new)
    if abs(x_new - x) < precision:   # converged
        break
    x = x_new

print(f"Iterations     : {i + 1}")
print(f"Minimum at x   = {x_new:.4f}")
print(f"Minimum value y = {f(x_new):.6f}")
```

### Step 3: Plot the descent

```python
xs = [v / 10 for v in range(-100, 41)]          # x from -10 to 4
plt.plot(xs, [f(v) for v in xs], label='y = (x+3)²')
plt.scatter(path, [f(v) for v in path], color='red', label='steps')
plt.xlabel('x'); plt.ylabel('y'); plt.legend()
plt.title('Gradient Descent'); plt.show()
```
